```python
import torch
from transformers import pipeline

device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"

pipe = pipeline(
    "automatic-speech-recognition",
    model="Kotib/uzbek_stt_v1",
    chunk_length_s=30,
    device=device
)

result = pipe("../../data/test/test_audio.mp3", generate_kwargs={"language": "uz", "task": "transcribe"})
print(result["text"])
```

```python
from pathlib import Path

transcription_path = Path("../../data/test/transcription.txt")
transcription_path.parent.mkdir(parents=True, exist_ok=True)
transcription_path.write_text(result["text"], encoding="utf-8")
```
