```python
from pathlib import Path
import nemo.collections.asr as nemo_asr

# Project root
ROOT = Path("/home/intern/book_audio_alignment")

# First audio section
audio_path = ROOT / (
    "src/notebooks/temporary_sections/section_000.wav"
)

# Load NVIDIA Uzbek model
model = nemo_asr.models.EncDecHybridRNNTCTCBPEModel.from_pretrained(
    model_name="nvidia/stt_uz_fastconformer_hybrid_large_pc",
    map_location="cpu"
)

# Use CTC decoding
model.change_decoding_strategy(decoder_type="ctc")

print("NeMo model loaded!")
```

```python
import soundfile as sf

# Read audio information
audio_info = sf.info(str(audio_path))

print("Sample rate:", audio_info.samplerate)
print("Channels:", audio_info.channels)
print("Duration:", round(audio_info.duration, 2), "seconds")

# Check compatibility
if audio_info.samplerate == 16000 and audio_info.channels == 1:
    print("Audio is ready for NeMo!")
else:
    print("Audio needs conversion before transcription.")
```

```python

# Use our previously loaded NeMo model
predictions = model.transcribe(
    [str(audio_path)],
    batch_size=1
)

# Extract the recognized text
result = predictions[0]

transcription = (
    result.text if hasattr(result, "text") else str(result)
)

print("TRANSCRIPTION:")
print(transcription)
```

```python

import json

# Load our previous matching results
match_path = ROOT / "data/test/book_matches_draft.json"

matches = json.loads(
    match_path.read_text(encoding="utf-8")
)

# Inspect the data structure
print("Data type:", type(matches).__name__)

if isinstance(matches, dict):
    print("Keys:", list(matches.keys()))
    print("First example:", str(list(matches.items())[:1])[:1000])

elif isinstance(matches, list):
    print("Total matches:", len(matches))
    print("First example:", str(matches[0])[:1000] if matches else "Empty")
```

```python

# Find our first section's matching result
match = next(
    item for item in matches
    if item["section"] == "section_000"
)

# Use the earlier transcription as our reference
reference_text = match["transcription"].strip()

# Prepare the manifest entry
entry = {
    "audio_filepath": str(audio_path.resolve()),
    "text": reference_text
}

# Update the manifest we created earlier
manifest_path = ROOT / "data/test/nemo_manifest_section_000.jsonl"

manifest_path.write_text(
    json.dumps(entry, ensure_ascii=False) + "\n",
    encoding="utf-8"
)

print("Manifest saved:", manifest_path)
print("Number of words:", len(reference_text.split()))
print("\nReference text:")
print(reference_text)
```

```python
import torch
from nemo.collections.asr.parts.utils.aligner_utils import (
    get_batch_variables,
    viterbi_decoding,
    add_t_start_end_to_utt_obj,
    Segment,
    Word,
)

model.eval()

# 1. Obtain model output for our audio and reference text
(
    log_probs,
    token_ids,
    audio_lengths,
    token_lengths,
    utterances,
    frame_duration,
) = get_batch_variables(
    audio=[str(audio_path)],
    model=model,
    gt_text_batch=[reference_text],
    align_using_pred_text=False,
)

# 2. Find where the reference tokens occur in the audio
alignments = viterbi_decoding(
    log_probs,
    token_ids,
    audio_lengths,
    token_lengths,
    viterbi_device=torch.device("cpu"),
)

# 3. Convert the alignment into timestamps
utterance = add_t_start_end_to_utt_obj(
    utterances[0],
    alignments[0],
    frame_duration,
)

# 4. Extract word-level timestamps
aligned_words = []

for segment in utterance.segments_and_tokens:
    if isinstance(segment, Segment):
        for word in segment.words_and_tokens:
            if isinstance(word, Word):
                aligned_words.append({
                    "word": word.text,
                    "start": word.t_start,
                    "end": word.t_end,
                })

# 5. Display the results
print("Total aligned words:", len(aligned_words))

for item in aligned_words[:15]:
    print(item)
```
