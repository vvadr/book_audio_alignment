```python
import librosa
from transformers import pipeline
```

```python
asr = pipeline(
    "automatic-speech-recognition",
    model="navai-uz/whisper-medium-uzbek",
    chunk_length_s=30,        # long-form audio is windowed automatically
)
```

```python
result = asr(
    "../../data/test/test_audio.mp3",
    generate_kwargs={"language": "uz", "task": "transcribe"}
)
```

```python
from pathlib import Path

transcription_path = Path("../../data/test/transcription.txt")
transcription_path.parent.mkdir(parents=True, exist_ok=True)
transcription_path.write_text(result["text"], encoding="utf-8")
```
